## 2. Merge

### a. Merge Basics

```python
# Read in spreadsheet
sales_may = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=0)
sales_june = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=1)
regions = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=2)
```

```python
sales_may
```

```python
regions
```

```python
# Accept defaults
sales_may.merge(regions)
```

What's happening here is inner join.
It joins by common cloumn of `store` but only includes stores that are in both tables.
But I want to keep all the rows in the left table.

```python
# Specify join
sales_may.merge(regions, how='left')
```

```python
# Specify left and right join columns
sales_may.merge(regions, how='left', left_on='store', right_on='store')
```

*   `left_on` and `right_on` arguments are more useful when column we want to join on have different names.
*   Use `on` argument only if column we want to join on have same names.
*   It is also possible to join to multiple columns. If tables have two columns with common values, specify both in `left_on` and `right_on` arguments as a list.


```python
# Use on if columns have same name
sales_may.merge(regions, how='left', on='store')

# More efficient code
```
