### a. Merge Basics

```python
# Read in spreadsheet
sales_may = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=0)
sales_june = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=1)
regions = pd.read_excel('/content/drive/MyDrive/Data Science in Pyhton by Maven Analytics/1. Data Prep & EDA/Data/Sales Tables.xlsx', sheet_name=2)
```

```python
sales_may
```

```python
# Accept defaults
sales_may.merge(regions)
```

What's happening here is ***inner*** join.
It joins by common cloumn of `store` but only includes stores that are in both tables.
But I want to keep all the rows in the left table.

```python
# Specify join
sales_may.merge(regions, how='left')
```

```python
# Specify left and right join columns
sales_may.merge(regions, how='left', left_on='store', right_on='store')
```

*   `left_on` and `right_on` arguments are more useful when column we want to join on have different names.
*   Use `on` argument only if column we want to join on have same names.
*   It is also possible to join to multiple columns. If tables have two columns with common values, specify both in `left_on` and `right_on` arguments as a list.


```python
# Use on if columns have same name
sales_may.merge(regions, how='left', on='store')

# More efficient code
```

## 4. Prepare Rows for Modeling

```python
# Goal: Predict which customers are most likely to buy dog food in June
## Each row should contain data for one customer
## y: June dog food purchases
## x: Aggregation based on April and May data
```

```python
# Check data types
df.dtypes
```

```python
# Convert data types to datetime and numeric fields
df.purchase_date = pd.to_datetime(df.purchase_date)
df.price = pd.to_numeric(df.price.str.replace('$', ''))
df.rating = pd.to_numeric(df.rating)
```

```python
# Check updated data types
df.dtypes
```

```python
# Create a subset of April and May data
df_april_may = df[df.purchase_date.dt.month < 6]
df_april_may.head()
```

```python
# create a subset of June data
df_june = df[df.purchase_date.dt.month == 6]
df_june.head()
```

```python
# Create a column of june dog food purchases
dog_food_rows = df_june[df_june.item_description == 'Dog Food']
dog_food_rows
```

```python
june_dog_food_purchases = (dog_food_rows
                           .groupby('customer')['item_id']
                           .count()
                           .rename('june_dog_food_purchases'))
june_dog_food_purchases
```

```python
# How much did each customer spend in April and May?
total_spend = (df_april_may
               .groupby('customer')['price']
               .sum()
               .rename('total_spend'))
total_spend
```

```python
# Concat june_dog_food_purchases and total_spend horizonatally
# Fill any NaN values with 0
model_df = (pd.concat([june_dog_food_purchases, total_spend], axis=1)
            .fillna(0)
            .reset_index())
model_df

# Each row now represents a customer. 'model_df' is ready for modeling!
```

## 5. Prepare Columns for Modeling

### a. Text to Numeric (Dummy Variables)

```python
# Goal: change the category column into numeric values
```

```python
df_april_may.head()
```

```python
# Specify a column to get dummy variables
category_dummies = pd.get_dummies(df_april_may.category)
category_dummies
```

```python
# Combine with customer data
pd.concat([df_april_may.customer, category_dummies], axis=1)
```

```python
# group by customer so each row is a single customer
categories = (pd.concat([df_april_may.customer, category_dummies], axis=1)
    .groupby('customer')
    .sum()
    .reset_index())
categories.head()
```

```python
model_df = model_df.merge(categories, how='left', on='customer')
model_df.head()
```

### b. DateTime to Numeric

```python
model_df.shape
```

```python
df_april_may.shape
```

```python
# Goal: add on a new column based on date (number of days since last purchase)
```

```python
# Find the last purchase date
last_purchase = df_april_may.groupby('customer')['purchase_date'].max()
last_purchase.head()
```

```python
# Note down today's date
today = pd.Series(pd.to_datetime('2023-06-01'), index=last_purchase.index)
today.head()
```

```python
# Find the number of days since the last purchase
days_between = (today - last_purchase).dt.days.rename('days_between')
days_between.head()
```

```python
days_between.reset_index()
```

```python
model_df = model_df.merge(days_between, how='left', on='customer')
model_df.head()
```
